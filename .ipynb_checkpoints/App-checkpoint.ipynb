```python
import streamlit as st
import pandas as pd
import glob
import os

st.set_page_config(page_title="LinkedIn Market Intelligence", layout="wide")

# Automatically grab the latest scraped CSV
csv_files = glob.glob("linkedin_*.csv")
if not csv_files:
    st.error("No scraped CSV files found in the current directory.")
    st.stop()

latest_file = max(csv_files, key=os.path.getctime)
df = pd.read_csv(latest_file)

# Sidebar Controls
st.sidebar.header("Filters")
selected_company = st.sidebar.multiselect("Company", options=df['Company'].unique())
selected_loc = st.sidebar.multiselect("Location", options=df['Location'].unique())

filtered_df = df.copy()
if selected_company:
    filtered_df = filtered_df[filtered_df['Company'].isin(selected_company)]
if selected_loc:
    filtered_df = filtered_df[filtered_df['Location'].isin(selected_loc)]

# Key Metrics
st.title(" Job Market Tracker")
st.caption(f"Loaded dataset: `{latest_file}`")

c1, c2, c3 = st.columns(3)
c1.metric("Total Jobs Extracted", len(filtered_df))
c2.metric("Unique Companies", filtered_df['Company'].nunique())
c3.metric("Remote Roles", filtered_df['Location'].str.contains('Remote', case=False).sum())

# Visuals
col1, col2 = st.columns(2)
with col1:
    st.subheader("Top Employers")
    st.bar_chart(filtered_df['Company'].value_counts().head(10))

with col2:
    st.subheader("Top Locations")
    st.bar_chart(filtered_df['Location'].value_counts().head(10))

# Interactive Raw Data Explorer
st.subheader("Explore Postings")
st.dataframe(
    filtered_df,
    column_config={"Link": st.column_config.LinkColumn("Posting Link")},
    use_container_width=True,
    hide_index=True
)
```
